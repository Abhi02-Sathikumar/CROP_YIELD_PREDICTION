```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
crop= pd.read_csv("/content/crop_production.csv")
crop
```

```python
print("Rows:", crop.shape[0])
print("Columns:", crop.shape[1])
```

```python
crop.columns = crop.columns.str.strip()
print(crop.columns.tolist())
```

```python
text_columns = [
    "State_Name",
    "District_Name",
    "Season",
    "Crop"
]

for col in text_columns:
    crop[col] = crop[col].astype(str).str.strip()
```

```python
crop.info()
```

```python
crop.describe()
```

```python
missing_values = crop.isnull().sum()

print("Missing values:")
print(missing_values)
```

```python
missing_percentage = (crop.isnull().sum() / len(crop)*100).sort_values(ascending=False)

missing_table = pd.DataFrame({
    "Missing_Count": crop.isnull().sum(),
    "Missing_Percentage": missing_percentage
})

missing_table
```

```python
crop["Season"].unique()
```

```python
print("Duplicate rows:", crop.duplicated().sum())
```

```python
print("Number of unique crops:", crop["Crop"].nunique())

print("\nCrop names:")
print(sorted(crop["Crop"].unique()))
```

```python
print("Number of districts:", crop["District_Name"].nunique())
```

```python
print("Missing Area:", crop["Area"].isna().sum())
print("Zero Area:", (crop["Area"] == 0).sum())
print("Negative Area:", (crop["Area"] < 0).sum())
```

```python
print("Minimum year:", crop["Crop_Year"].min())
print("Maximum year:", crop["Crop_Year"].max())

print("\nYears:")
print(sorted(crop["Crop_Year"].unique()))
```

```python
print(crop.dtypes)
```

```python
crop = crop[crop["Area"] > 0].copy()

print("Shape after Area validation:", crop.shape)
```

```python
print("Missing Production:", crop["Production"].isna().sum())
print("Zero Production:", (crop["Production"] == 0).sum())
print("Negative Production:", (crop["Production"] < 0).sum())
```

```python
crop = crop.dropna(subset=["Production"]).copy()

print("Shape after removing missing Production:", crop.shape)
```

```python
crop["Yield"] = crop["Production"] / crop["Area"]

crop[["Area", "Production", "Yield"]].head()
```

```python
print("Missing Yield:", crop["Yield"].isna().sum())
print("Infinite Yield:", np.isinf(crop["Yield"]).sum())
print("Negative Yield:", (crop["Yield"] < 0).sum())
```

```python
print("Final shape:", crop.shape)

print("\nMissing values:")
print(crop.isnull().sum())

print("\nDuplicate rows:")
print(crop.duplicated().sum())

print("\nData types:")
print(crop.dtypes)
```
