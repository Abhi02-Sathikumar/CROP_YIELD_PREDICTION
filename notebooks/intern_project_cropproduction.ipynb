```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load the raw crop production dataset from GitHub

url = "https://raw.githubusercontent.com/Abhi02-Sathikumar/CROP_YIELD_PREDICTION/main/data/raw/crop_production.csv"
crop = pd.read_csv(url)

print("Dataset loaded successfully!")

crop.head()
```

```python
print("Rows:", crop.shape[0])
print("Columns:", crop.shape[1])
```

```python
crop.columns = crop.columns.str.strip()
print(crop.columns.tolist())
```

```python
text_columns = [
    "State_Name",
    "District_Name",
    "Season",
    "Crop"
]

for col in text_columns:
    crop[col] = crop[col].astype(str).str.strip()
```

```python
missing_values = crop.isnull().sum()

print("Missing values:")
print(missing_values)
```

```python
missing_percentage = (crop.isnull().sum() / len(crop)*100).sort_values(ascending=False)

missing_table = pd.DataFrame({
    "Missing_Count": crop.isnull().sum(),
    "Missing_Percentage": missing_percentage
})

missing_table
```

```python
print("Number of unique crops:", crop["Crop"].nunique())

print("\nCrop names:")
print(sorted(crop["Crop"].unique()))
```

```python
print(crop.dtypes)
```

```python
crop = crop[crop["Area"] > 0].copy()

print("Shape after Area validation:", crop.shape)
```

```python
print("Missing Production:", crop["Production"].isna().sum())
print("Zero Production:", (crop["Production"] == 0).sum())
print("Negative Production:", (crop["Production"] < 0).sum())
```

```python
crop = crop.dropna(subset=["Production"]).copy()

print("Shape after removing missing Production:", crop.shape)
```

```python
crop["Production"].isna().sum()
```

```python
crop["Yield"] = crop["Production"] / crop["Area"]

crop[["Area", "Production", "Yield"]].head()
```

```python
print("Missing Yield:", crop["Yield"].isna().sum())
print("Infinite Yield:", np.isinf(crop["Yield"]).sum())
print("Negative Yield:", (crop["Yield"] < 0).sum())
```

```python
print("Final shape:", crop.shape)

print("\nMissing values:")
print(crop.isnull().sum())

print("\nDuplicate rows:")
print(crop.duplicated().sum())

print("\nData types:")
print(crop.dtypes)
```

```python
import pandas as pd

# Use your cleaned dataset
map_data = crop.dropna(
    subset=["State_Name", "Production", "Area"]
).copy()

map_data = map_data[map_data["Area"] > 0]

# Calculate yield
map_data["Yield"] = map_data["Production"] / map_data["Area"]

# Aggregate across all crops and years
state_yield = (
    map_data.groupby("State_Name")
    .agg(
        Production=("Production", "sum"),
        Area=("Area", "sum")
    )
    .reset_index()
)

state_yield["Yield"] = (
    state_yield["Production"] / state_yield["Area"]
)

state_yield.head()
```

```python
# Match crop dataset names to GeoJSON names
state_yield["State_Name"] = state_yield["State_Name"].replace({
    "Andaman and Nicobar Islands": "Andaman and Nicobar",
    "Odisha": "Orissa",
    "Uttarakhand": "Uttaranchal"
})
```

```python
import plotly.express as px

fig = px.choropleth(
    state_yield,
    geojson=india_geojson,
    locations="State_Name",
    featureidkey="properties.NAME_1",
    color="Yield",
    hover_name="State_Name",
    hover_data={
        "Production": ":,.0f",
        "Area": ":,.0f",
        "Yield": ":.2f"
    },
    color_continuous_scale="YlGn",
    title="Overall Crop Production per Unit Area by Indian State"
)

fig.update_geos(
    fitbounds="locations",
    visible=False
)

fig.update_layout(
    margin={"r": 0, "t": 50, "l": 0, "b": 0}
)

fig.show()
```
